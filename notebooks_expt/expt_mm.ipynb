```python
import torch
from torchvision import transforms as T
```

```python
def get_image(a):
    return T.ToPILImage()(a)

img_size = 128
a = torch.ones(img_size,img_size)
img_1 = torch.nn.functional.pad(torch.triu(a)/5,(16,16,16,16),value=1)
img_2 = torch.nn.functional.pad(torch.tril(a)/5,(16,16,16,16),value=1)
img_3 = img_1@img_2
img_4 = img_2@img_1
img_5 = (img_1+img_2)/2

imgs = torch.concat([img_1,img_2,img_3,img_4,img_5],dim=1)
display(get_image(imgs))
# for img in imgs:
#     display(get_image(img))
```
