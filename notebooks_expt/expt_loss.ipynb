```python
import torch
import torch.nn as nn
```

```python
a1 = torch.randn(8,1025,1024)
a2 = torch.randn(8,1025,1024)

def dot(x,y):
    x = torch.mean(x,dim=-1)
    y = torch.mean(y,dim=-1)
    return torch.mean((x*y).sum(dim=-1))

def get_loss(y_t,y_p):
    loss_1 = dot(y_t,y_p)
    loss_2 = dot(y_t.transpose(-1,-2),y_p.transpose(-1,-2))
    return loss_1 + loss_2

loss = get_loss(a1,a2)
print(loss)
```

```python
loss_fn = torch.nn.MSELoss()
```

```python
loss_a = 0
for i in range(1025):
    loss_a = loss_a + loss_fn(a1[:,0,:],a2[:,0,:])
```

```python
loss_b = loss_fn(a1,a2)
```

```python
loss_a/1025,loss_b
```
