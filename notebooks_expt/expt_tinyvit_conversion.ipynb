```python
import torch
import timm
```

```python
model = timm.create_model('tiny_vit_21m_512.dist_in22k_ft_in1k', pretrained=False, features_only=True)
model = model.eval()
```

```python
class TinyViT(torch.nn.Module):
    def __init__(self,backbone):
        super().__init__()
        self.bb = backbone
    def forward(self,x):
        return self.bb(x)[1:]

model = TinyViT(model)
```

```python
x = model(torch.randn(1,3,512,512))
x[0].shape,x[1].shape,x[2].shape
```

```python
torch.onnx.export(
    model,
    torch.randn(1,3,512,512),
    "tiny_vit.onnx",
    input_names=["image"],
    output_names=["out_0","out_1","out_2"]
)
```
